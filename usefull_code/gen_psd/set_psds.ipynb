### Info
We take the public ET psds from https://gitlab.et-gw.eu/pub/et-sensitivity-curves/-/tree/main?ref_type=heads.

We consider all the combined noise contribution. We take the 'strain displacement' data from column 2, and we rescale by length and combine HF and LF.

```python
# these need to be scaled by the arm lenght (in meters!)
new10kmHF = pd.read_csv('psds/' + 'total_10HF.txt', names=['Frequency', 'ASD_HF', 'ASD_other1', 'ASD_other2'], delimiter=' ', skiprows=0)
new15kmHF = pd.read_csv('psds/' + 'total_15HF.txt', names=['Frequency', 'ASD_HF', 'ASD_other1', 'ASD_other2'], delimiter=' ', skiprows=0)
new10kmLF = pd.read_csv('psds/' + 'total_10LF.txt', names=['Frequency', 'ASD_LF', 'ASD_other1', 'ASD_other2'], delimiter=' ', skiprows=0)
new15kmLF = pd.read_csv('psds/' + 'total_15LF.txt', names=['Frequency', 'ASD_LF', 'ASD_other1', 'ASD_other2'], delimiter=' ', skiprows=0)
```

```python
#rescale by arm length
ASD_10kmHF = (new10kmHF['ASD_HF']/10_000.)
ASD_15kmHF = (new15kmHF['ASD_HF']/15_000.)
ASD_10kmLF = (new10kmLF['ASD_LF']/10_000.)
ASD_15kmLF = (new15kmLF['ASD_LF']/15_000.)
ASD10 = 1./(1./(new10kmHF['ASD_HF']/10_000.) + 1./(new10kmLF['ASD_LF']/10_000.))
ASD15 = 1./(1./(new15kmHF['ASD_HF']/15_000.) + 1./(new15kmLF['ASD_LF']/15_000.))
```

```python
#rescale by arm lenght individually and convert to psd
PSD_10kmHF = 1./(1./(new10kmHF['ASD_HF']/10_000.)**2 )
PSD_15kmHF = 1./(1./(new15kmHF['ASD_HF']/15_000.)**2 )
PSD_10kmLF = 1./(1./(new10kmLF['ASD_LF']/10_000.)**2 )
PSD_15kmLF = 1./(1./(new15kmLF['ASD_LF']/15_000.)**2 )
```

```python
# rescale by arm length and combine
PSD10 = 1./(1./(new10kmHF['ASD_HF']/10_000.)**2 + 1./(new10kmLF['ASD_LF']/10_000.)**2)
PSD15 = 1./(1./(new15kmHF['ASD_HF']/15_000.)**2 + 1./(new15kmLF['ASD_LF']/15_000.)**2)
```

```python
#save as asd files
et10ASD= np.c_[new10kmHF['Frequency'],ASD10]
et15ASD= np.c_[new15kmHF['Frequency'],ASD15]

et10ASD_LF= np.c_[new10kmLF['Frequency'],ASD_10kmLF]
et15ASD_LF= np.c_[new15kmLF['Frequency'],ASD_15kmLF]
et10ASD_HF= np.c_[new10kmHF['Frequency'],ASD_10kmHF]
et15ASD_HF= np.c_[new15kmHF['Frequency'],ASD_15kmHF]

np.savetxt('asds/et_10fullcryo_asd.txt', et10ASD)
np.savetxt('asds/et_15fullcryo_asd.txt', et15ASD)
np.savetxt('asds/et_10_asd_LF.txt', et10ASD_LF)
np.savetxt('asds/et_15_asd_LF.txt', et15ASD_LF)
np.savetxt('asds/et_10_asd_HF.txt', et10ASD_HF)
np.savetxt('asds/et_15_asd_HF.txt', et15ASD_HF)
```

```python
# save as psd files
et10PSD = np.c_[new10kmHF['Frequency'], PSD10]
et15PSD = np.c_[new15kmHF['Frequency'], PSD15]

et10PSD_LF = np.c_[new10kmLF['Frequency'], PSD_10kmLF]
et15PSD_LF = np.c_[new15kmLF['Frequency'], PSD_15kmLF]
et10PSD_HF = np.c_[new10kmHF['Frequency'], PSD_10kmHF]
et15PSD_HF = np.c_[new15kmHF['Frequency'], PSD_15kmHF]

np.savetxt('psds/et_10fullcryo_psd.txt', et10PSD)
np.savetxt('psds/et_15fullcryo_psd.txt', et15PSD)
np.savetxt('psds/et_10_psd_LF.txt', et10PSD_LF)
np.savetxt('psds/et_15_psd_LF.txt', et15PSD_LF)
np.savetxt('psds/et_10_psd_HF.txt', et10PSD_HF)
np.savetxt('psds/et_15_psd_HF.txt', et15PSD_HF)
```
