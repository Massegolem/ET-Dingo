```python
import numpy as np
import matplotlib.pyplot as plt

asd_files = {
    "aLIGO O4 high": "aLIGO_O4_high_asd.txt",
    "aLIGO ZERO_DET high P": "aLIGO_ZERO_DET_high_P_asd.txt",
    "ligo_O3": "ligo_O3.txt",
}

data = {}
for label, path in asd_files.items():
    freq, asd = np.loadtxt(path, unpack=True)
    data[label] = (freq, asd)
    print(f"{label}: {freq.size} points, {freq.min():.2f}-{freq.max():.2f} Hz")
```

```python
fig, ax = plt.subplots(figsize=(8, 6))

for label, (freq, asd) in data.items():
    ax.loglog(freq, asd, label=label, lw=1.5)

ax.set_xlabel("Frequency [Hz]")
ax.set_ylabel(r"ASD [1/$\sqrt{\mathrm{Hz}}$]")
ax.set_title("Amplitude Spectral Density Comparison")
ax.grid(True, which="both", ls="--", alpha=0.4)
ax.legend()
fig.tight_layout()
plt.show()
```
